```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
'''Report_ID: String (e.g., "SOS_1" to "SOS_1000")

Latitude: Float (26.8000 to 27.0000 - Jaipur limits)

Longitude: Float (75.7000 to 75.9000 - Jaipur limits)

Zone: String (Categorical: "North", "South", "East", "West")

Total_Trapped: Integer (10 to 1,000 people)

Children_Trapped: Integer (2 to 350 people; approx 20-35% of Total_Trapped)

Seniors_Trapped: Integer (0 to 150 people; approx 5-15% of Total_Trapped)

Severity_Level: Integer (1 to 5)

Infra_Damage_Pct: Integer (0 to 100 percentage points)

Temperature_C: Integer (28 to 48 degrees Celsius)

Water_Liters_Needed: Integer (30 to 4,500+ liters)

_Kits_Needed: Integer (0 to 150+ kits)'''
```

```python
Zones=np.random.choice(['North','South','East','West'],1000)
```

```python
Total_Trapped=np.random.randint(10,1001,1000)
```

```python
children_trapped=(Total_Trapped*np.random.uniform(0.20,0.35)).astype(int)
```

```python
senior_trapped=(Total_Trapped*np.random.uniform(0.05,0.15)).astype(int)
```

```python
severity_level=np.random.choice([1,2,3,4,5],1000)
```

```python
bounds = {
    "North": {"lat": (26.9300, 27.0000), "lon": (75.7500, 75.8500)},
    "South": {"lat": (26.8000, 26.8800), "lon": (75.7500, 75.8500)},
    "East":  {"lat": (26.8900, 26.9500), "lon": (75.8200, 75.9000)},
    "West":  {"lat": (26.8900, 26.9500), "lon": (75.7000, 75.7700)}
}
```

```python
latitudes = np.array([np.random.uniform(*bounds[z]["lat"]) for z in Zones])
longitudes = np.array([np.random.uniform(*bounds[z]["lon"]) for z in Zones])
```

```python
report_id=np.array([f"SOS_{i}" for i in range(1,1001)])
```

```python
infra_damage_pct=np.random.randint(0,101,1000)
```

```python
temperature_C=np.random.randint(28,49,1000)
```

```python
temp_multiplier=np.where(temperature_C>40,1.5,1.0)
```

```python
water_liters_needed=(Total_Trapped*3*temp_multiplier)
```

```python
kits_needed=((severity_level*5)+(children_trapped*0.2)+(senior_trapped*0.3)).astype(int)
```

```python
report={"Report_id":report_id,"Latitude":latitudes ,"Longitude":longitudes,"Zone":Zones,"Total_Trapped":Total_Trapped,"Children_Trapped":children_trapped,"Seniors_Trapped":senior_trapped,"Severity_Level":severity_level,"Infra_Damage_Pct":infra_damage_pct,"Temperatur_C":temperature_C,"Water_Liters_Needed":water_liters_needed,"Kits_Needed":kits_needed}
```

```python
df=pd.DataFrame(report)
```

```python
df.to_csv("sos_emergency_reports.csv",index=False)
```

```python
df1=df.copy()
```

```python
gov_water_stock=500000
kit_stock=25000
```

```python
water_status=[]
kit_status=[]
water_allocated=[]
kit_allocated=[]
```

```python
water_rem=gov_water_stock
kit_rem=kit_stock
```

```python
df1=df.sort_values(by=["Water_Liters_Needed",'Kits_Needed'],ascending=False,inplace=False)
for i in df1.Water_Liters_Needed:
  if water_rem>=i:
    water_rem=water_rem-i
    water_status.append("Available")
    water_allocated.append(i)
  elif water_rem>0:
    water_status.append("partially available")
    water_allocated.append(water_rem)
    water_rem=0
  else:
    water_status.append("Unavailable")
    water_allocated.append(0)

for j in df1.Kits_Needed:
  if kit_rem>=j:
    kit_rem=kit_rem-j
    kit_status.append("Available")
    kit_allocated.append(j)
  elif kit_rem>0:
    kit_status.append("partially available")
    kit_allocated.append(kit_rem)
    kit_rem=0
  else:
    kit_status.append("Unavailable")
    kit_allocated.append(0)
df1["water_status"]=water_status
df1["water_allocated"]=water_allocated
df1["kit_status"]=kit_status
df1["kit_allocated"]=kit_allocated
```

```python
df1
```

```python
df1["water_shortage"]=df1["Water_Liters_Needed"]-df1["water_allocated"]
df1["kit_shortage"]=df1["Kits_Needed"]-df1["kit_allocated"]
```
